```python
data_yaml = """
path: inspection.yolov12

train: images/train
# val: images/val

nc: 24
names: [
'roughness_1','roughness_10','roughness_11','roughness_12','roughness_13','roughness_14',
'roughness_15','roughness_16','roughness_17','roughness_18','roughness_19','roughness_2',
'roughness_20','roughness_21','roughness_22','roughness_23','roughness_24',
'roughness_3','roughness_4','roughness_5','roughness_6','roughness_7','roughness_8','roughness_9'
]
"""

with open("data.yaml", "w") as f:
    f.write(data_yaml)
```

```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")   # start small
```

```python
model.train(
    data="inspection.yolov12/data.yaml",
    epochs=10,
    imgsz=640,
    batch=16,
    name="steel_detectorv3"
)
```

```python
metrics = model.val()
print(metrics)
```

```python
from pprint import pprint

print("Model validation summary:")
print("type:", type(metrics))
for name in [
    "box", "cls", "dfl", "mAP50", "mAP50_95", "map50", "map50_95", "p", "r", "obj",
    "box_loss", "cls_loss", "dfl_loss", "fitness"
]:
    if hasattr(metrics, name):
        print(f"{name}:", getattr(metrics, name))

print("\nAvailable metric fields:")
print([n for n in dir(metrics) if not n.startswith("_")])
```

```python
from PIL import Image
from IPython.display import display
from ultralytics import YOLO

model = YOLO("best.pt")

print("✅ Model loaded")

def test_image(img_path, conf=0.015):
    results = model(img_path, conf=conf)

    # Plot result
    result_img = results[0].plot()

    display(Image.fromarray(result_img))

    # Print detections
    print("\nDetections:")
    for box in results[0].boxes:
        cls = int(box.cls[0])
        conf = float(box.conf[0])
        print(f"Class: {model.names[cls]} | Confidence: {conf:.2f}")
```

```python
test_image("Data/5/5_5.jpeg")
```
